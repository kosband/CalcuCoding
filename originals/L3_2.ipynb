Helper functions carried forward from earlier parts. **plot_xy** adds an option to fill a figure.

```python
%pip install ipycc
from ipycc.turtle import Turtle, showscreen, clearscreen

def linear_spacing(xmin, xmax, num):
    interval = (xmax - xmin) / (num - 1)
    xlist = [xmin + i * interval for i in range(num)]
    xlist[-1] = xmax    # guarantees tha last element exactly equals xmax     
    return xlist
 
def plot_xy(x, y, color='black', fill=False):
    t = Turtle()
    t.speed('fastest')
    t.penup()
    t.goto(x[0], y[0])
    t.color(color)
    t.pendown()
    if fill: t.begin_fill()
    for i in range(1, len(x)):
        t.goto(x[i], y[i])
    if fill: t.end_fill()
    t.hideturtle()
```

Let's chart conic sections using turtle. Below we plot a circle using square roots to identify (*x*,*y*) pairs.

```python
from math import sqrt  # imports only sqrt from math, which we use without "math." prefix

def circle(radius, num):
    x = linear_spacing(radius, -radius, num)                      # x progresses right to left
    y = [sqrt(radius * radius - val * val) for val in x]          # heights of upper semicircle
    x_lower = linear_spacing(-radius, radius, num)                # x progresses left to right
    y_lower = [-sqrt(radius * radius - val * val) for val in x]   # heights of upper semicircle
    x = x + x_lower[1:]     # combines two lists in order after removing duplicate
    y = y + y_lower[1:]     # combines two lists in order after removing duplicate
    return x,y
      
# demo
showscreen()
clearscreen()
x,y = circle(200,40); plot_xy(x, y, 'saddlebrown')  
x,y = circle(100,40); plot_xy(x, y, 'brown')  
x,y = circle(50,80); plot_xy(x, y, 'chocolate', True)  
x,y = circle(25,120); plot_xy(x, y, 'gold', True)  
```

The previous plots are jagged when the ratio of num to radius is low and slow to form when num is high. 
Also, many of the *x* and *y* computations are redundant. Let's make some simple modifications.

```python
def circle(radius, num=None):
    if num==None:
        num = 20 + radius//3
    x = linear_spacing(radius, -radius, num)
    y = [sqrt(radius * radius - val * val) for val in x]
    x_lower = x[-1::-1]             # lists x in reversed order  
    y_lower = [-val for val in y]   # list of -y
    x.extend(x_lower[1:])  # alternative to x = x + x_lower[1:]
    y.extend(y_lower[1:])
    return x,y
    
# demo  
showscreen()
clearscreen()
x,y = circle(200); plot_xy(x, y, 'plum')  
x,y = circle(100); plot_xy(x, y, 'purple')  
```

The previous code drew two ellipses. That makes sense, 
because *x*=*aX* and *y*=*bY* given *X*<sup>2</sup> + *Y*<sup>2</sup> = 1
imply the ellipse equation (*x/a*)<sup>2</sup> + (*y/b*)<sup>2</sup> = 1.

```python
def ellipse(a, b):
    num = 20 + max(a,b)//3
    x,y = circle(1, num)
    xnew = [a * val for val in x]
    ynew = [b * val for val in y]
    return xnew,ynew

showscreen()
clearscreen()
x,y = ellipse(150, 100)
plot_xy(x, y, 'turquoise', True)
```

One problem with our approach is that we're calculating and returning the *x* and *y* lists separately 
when in fact each *x*[i],*y*[i] pair is closely linked. 
One way to address this uses the python command **zip**(*x,y*), which iterates through (*x*[i],*y*[i]) pairs. 
Here is a simple example.

```python
x = [0,1,2,3,4]
y = [2 * val for val in x]
z = zip(x,y)
print('values as iterate through z=zip(x,y)')
for val in z:
    print(val)      # iterate over x[i],y[i] pairs
print('\nvalues on second iteration through z:') # print '\n' starts new line   
for val in z:       # repeating this does nothing as each element forgotten after first iteration 
    print(val,' ')
print('\nlist of first iterations through zip(x,y)')    
z_list = list(zip(x,y)) # convert zip iteration into list, similiar to list(range()) in that respect    
print(z_list)
```

Let's rewrite **ellipse** to use **zip** and also rewrite **plot_xy** to accept paired values. 
To unzip a zipped sequence *z*, use the command **zip**(\**z*).

```python
def ellipse(a,b):
    num = 20 + (a + b)//6
    x,y = circle(1, num)
    # generate list of coordinate tuples
    coords = [(a*val[0],b*val[1]) for val in zip(x,y)]
    return coords

def plot_xy(x, y=None, color='black', fill=False):
    t = Turtle()
    t.speed('fastest')
    if type(y) is str:  # True if and only if y is a string
        color = y       # reassign y to color
        y = None        # reassign None to y
    if y==None:         # if True, x must be zipped file
        x,y = zip(*x)   # unpacks x into separate x and y tuples
    t.color(color)
    t.penup()
    t.goto(x[0], y[0])
    t.pendown()
    if fill: t.begin_fill()
    for i in range(1, len(x)):
        t.goto(x[i], y[i])
    if fill: t.end_fill()
    t.hideturtle()

showscreen()
clearscreen()
coords = ellipse(50,100)
plot_xy(coords,'skyblue')
x,y = zip(*coords)
print('Both x and y are',type(x),'with length',len(x),'; x[0]=',x[0])
plot_xy(y,x)              # here inputs are two unzipped sequences
#t.hideturtle()
```

Each focus of an ellipse lies *c* = sqrt(*a*<sup>2</sup> - *b*<sup>2</sup>) from the origin along the longest axis. 
This code maps the foci, draws line segments from the foci to a chosen point on the periphery, 
and confirms that the distances sum to max(2*a*,2*b*).

```python
def distance_between(point1,point2):
    return sqrt((point2[0]-point1[0])**2 + (point2[1]-point1[1])**2)

def plot_foci_to_point(focus1,focus2,point):
    # multiple commands can be combined on one line if separated by semi-colon
    # generally discouraged, but occasionally makes logic clearer and code easier to read
    t = Turtle()
    t.color('black'); t.hideturtle(); t.penup()
    t.goto(focus1); t.pendown(); t.dot(); t.write('focus1')
    t.goto(point); t.dot(); t.write('x,y')
    t.goto(focus2); t.dot(); t.write('focus2')

def plot_lines_to_ellipse_foci(a,b,x_val,negative_y=False):
    # compute foci
    c = sqrt(abs(a**2 - b**2))
    focus1 = ((c,0) if a>b else (0,c))
    focus2 = (-focus1[0],-focus1[1])
    # compute point on periphery and plot to foci
    x_val = min(a,max(x_val,-a))   # keep x_val between -a and +a to avoid error
    y_val = b/a * sqrt(a*a - x_val * x_val)
    if negative_y:
        y_val = -y_val
    point = (x_val,y_val)
    plot_foci_to_point(focus1,focus2,point)
    # measure and report sums of distances
    sum_distance = distance_between(focus1,point) + distance_between(focus2,point)
    deviation = abs(2 * max(a,b) - sum_distance)
    print('The distances sum to',sum_distance,', which is',deviation,'from expected')
    
showscreen()
clearscreen()
a,b,x_val=150,100,50  # try with different values
plot_xy(ellipse(a,b),'red')
plot_lines_to_ellipse_foci(a,b,x_val)
```

Next let's chart some parabolas. A parabola pointing up from the origin with focus *F* has
equation *y* = *x*<sup>2</sup>/(4*F*). Parabolas gets shallower as *F* rises.

```python
def parabola(x,focus,base):
    return [(1/4/focus) * val * val + base for val in x]

showscreen()
clearscreen()
x = linear_spacing(-200,200,101)  # equivalent to list(range(-200,201,2))
colors = ['red','orange','green','blue','indigo','violet']
base = -180
t = Turtle()
t.hideturtle()
for i in range(6):
    focus = 20 + 15 * i
    y = parabola(x,focus,base)
    plot_xy(x, y, colors[i])
    plot_xy([0,0],[base,base+focus],colors[i])
    t.color(colors[i]); t.penup()
    t.goto(0, base+focus); t.pendown()
    t.dot(10)     # makes dot of size 10
    t.write('   focus')
```

As explained in the text, we can easily rotate figures. This function codes the procedure sketched there.

```python
def rotation(x,y,slope,negative_mult=False):
    mult = 1 / sqrt(1 + slope * slope)
    if negative_mult: # this will rotate an additional 180 degrees
        mult = -mult   
    new_x = [mult * (x[i] - slope * y[i]) for i in range(len(x))]
    new_y = [mult * (slope * x[i] + y[i]) for i in range(len(x))]
    return zip(new_x,new_y)

# rotate parabolas
showscreen()
clearscreen()
x = linear_spacing(-200,200,101)
y = parabola(x,25,-100)
plot_xy(rotation(x,y,1), 'red')
plot_xy(rotation(x,y,-1), 'green')
plot_xy(rotation(x,y,-0.25), 'blue')
plot_xy(rotation(x,y,4), 'indigo')
plot_xy(rotation(x,y,0.2,True),'violet')
```

Next let's plot hyperbolas. 
Unlike with ellipses, there are no conceptual advantages to coding a unit hyperbola first.

```python
def hyperbola(a,b):
    x = linear_spacing(a,200,70) 
    y = [(b/a) * sqrt(val * val - a * a) for val in x]
    y_minus = [-val for val in y]    
    x_right = x[-1::-1] + x[1:]
    y_right = y[-1::-1] + y_minus[1:]
    branch_right = list(zip(x_right,y_right))  # forms list of tuple pairs x_right[i],y_right[i]
    branch_left = [(-z[0],z[1]) for z in branch_right]
    return branch_right,branch_left

showscreen()
clearscreen()
branch_right,branch_left = hyperbola(20,40)
plot_xy(branch_right, 'blue') # need to plot branches separately as they're not connected
plot_xy(branch_left, 'blue')
branch_right,branch_left = hyperbola(20,20)
plot_xy(branch_right, 'red')
plot_xy(branch_left, 'red')
branch_right,branch_left = hyperbola(40,40)
plot_xy(branch_right, 'green')
plot_xy(branch_left, 'green')
branch_right,branch_left = hyperbola(40,20)
plot_xy(branch_right, 'violet')
plot_xy(branch_left, 'violet')
```

Let's add plots of the asymptotes to the hyperbola branches.

```python
def plot_asymptotes_hyperbola(a,b,pen_color='black'):
    xmax = ymax = 200
    if a>b: ymax = (b/a) * ymax
    else:   xmax = (a/b) * ymax
    t = Turtle()
    t.speed('fastest')
    t.color(pen_color);  t.penup()
    t.goto(xmax,ymax);   t.pendown()
    t.goto(-xmax,-ymax); t.penup()
    t.goto(xmax,-ymax);  t.pendown()
    t.goto(-xmax,ymax)    

showscreen()
clearscreen()
a,b = 40,30
branch_right,branch_left = hyperbola(a,b)
plot_xy(branch_right,'blue')
plot_xy(branch_left,'blue')
plot_asymptotes_hyperbola(a,b,'red')
```

Each focus of a hyperbola lies *c* = sqrt(*a*<sup>2</sup> + *b*<sup>2</sup>) from the origin along the main axis. 
The code here maps the foci, draws line segments from the foci to a chosen point on the hyperbola, 
and confirms that the distances differ by 2*a*.

```python
def plot_lines_to_hyperbola_foci(a,b,x_val,negative_y=False):
    # compute foci
    c = sqrt(a * a + b * b)
    focus1 = (c,0)
    focus2 = (-c,0)
    # compute point on periphery and plot to foci
    if abs(x_val) < a: print('No point with x =', x_val, 'lies on the hyperbolas'); return
    y_val = b/a * sqrt(x_val * x_val - a * a)
    if negative_y: y_val = -y_val   # compressed to single line with no loss of clarity
    point = (x_val,y_val)
    plot_foci_to_point(focus1,focus2,point)
    # measure and report difference of distances
    diff_distance = abs(distance_between(focus1,point) - distance_between(focus2,point))
    deviation = abs(2 * a - diff_distance)
    print('The distances differ by',diff_distance,', which is',deviation,'from expected')
                        
showscreen()
x_val = -100
plot_lines_to_hyperbola_foci(a,b,x_val)
```
