```python
# initializations
%pip install ipycc
from ipycc.turtle import Turtle, showscreen, clearscreen
import numpy as np
import matplotlib.pyplot as plt

def plot_xy(x,y,pen_color='black'):
    t = Turtle()
    t.speed('fastest')
    t.color(pen_color)
    t.penup(); t.goto(x[0],y[0]); t.pendown()
    for i in range(1,len(x)):
        t.goto(x[i],y[i])
    t.hideturtle()
```

It's easy to plot in polar coordinates using matplotlib.pyplot. The baseline projection includes a grid with 5 concentric circles 
labeled with modulus and 8 radial lines labeled in degrees. The maximum displayed modulus will automatically adjust from its baseline of 1.

```python
# Be patient; this takes a while
plt.figure()
plt.axes(projection='polar')
plt.title('Baseline Polar Grid')  # must come after axes to avoid drawing rectangle

plt.figure()
theta = np.linspace(0,2*np.pi,100)
r = np.ones((len(theta),5))
for n in range(5):
    r[:,n] = 10*(n+1)
plt.polar(theta,r)
plt.title('Circles')

plt.figure()
theta = np.ones((100,5))
for i in range(5):
    theta[:,i] = i * 2 * np.pi / 5
r = np.linspace(0,len(theta),100)
plt.polar(theta,r)
plt.title('Radial Lines')
plt.show()
```

To rewrite the degrees in radian terms, 
we need to return a specific axis object--here called 'ax'--
and edit using the package **ticker** in matplotlib. 
There is no need to memorize this. 
It is included only to illustrate the potential sophistication of matplotlib.

```python
import matplotlib.ticker as ticker

def radians(x, y =None):
    rad_x = x / np.pi
    return "{}π".format(str(rad_x if rad_x % 1 else int(rad_x)))

ax = plt.subplot(111, projection='polar')
ax.xaxis.set_major_formatter(ticker.FuncFormatter(radians))
plt.title('Polar Grid in Radians')
plt.show()   # be patient
```

Often it's useful to turn off the polar grid and labels. 
Here is code that does that and plots filled concentric polygons.

```python
def polar_unlabeled():
    plt.axes(projection='polar')
    plt.grid(False)
    plt.xticks([ ])
    plt.yticks([ ])
    plt.box([ ])

polar_unlabeled()
plt.title('Filled Regular Polygons')
for sides in (12,10,8,6,5,4,3):
    theta = np.linspace(0,2*np.pi,sides+1)
    r = 10 * sides + np.zeros(sides+1)
    plt.fill(theta,r)
plt.show()
```

Polar plots can link a given *x* to multiple values of *y*. 
Consider for example a roll of wrapping paper, where each layer is separated from the next by a constant thickness. 
The mathematical counterpart is known as an arithmetic spiral, with equation *r* = *a* + *b*$\theta$.

```python
circuits,a,b = 8,1,1
theta = np.arange(0,circuits,0.01) * 2 * np.pi
r = a + b * theta
polar_unlabeled()
plt.title('Arithmetic Spiral')
plt.polar(theta,r)
plt.show()
```

The function *r = a* |sin n$\theta$ | makes "roses" with 2*n* petals.

```python
a = 1
theta = np.arange(0,2 * np.pi,0.01)
for n in (3,5,8):
    plt.figure()
    polar_unlabeled()
    plt.title('Rose with ' + str(2*n) + ' Petals') 
    r = a * np.abs(np.sin(n * theta))
    plt.plot(theta,r)  # works same as plt.polar once polar plot is specified
plt.show()
```

By shrinking *r* we can superimpose layers of rose-shaped petals.

```python
a,n = 1,6
theta = np.arange(0,2 * np.pi,0.01)
plt.figure()
polar_unlabeled()
plt.title('Rose with 3 Layers of ' + str(2*n) + ' Petals Each')
r = a * np.abs(np.sin(n * theta))
plt.polar(theta,r)
plt.fill(theta,r/2)
plt.fill(theta,r/4)
plt.show()
```

Substituting cos *n*$\theta$ for sin *n*$\theta$ rotates the flower by half a petal.

```python
a,n = 10,8
theta = np.arange(0,2 * np.pi,0.01)
polar_unlabeled()
plt.title('Roses with 2 Layers of ' + str(2*n) + ' Petals Each') 
r = a * np.abs(np.cos(n * theta))
plt.polar(theta,r)
plt.fill(theta,r)
r = a * np.abs(np.sin(n * theta))
plt.fill(theta,r)
plt.show()
```

Let's plot ellipses in polar coordinates. 
The simplest method places a focus at the origin 
and defines ellipses by their maximum radius *a* and their eccentricity.

```python
def ellipse(theta, a, eccentricity):
    return a * (1 - eccentricity**2) / (1 + eccentricity * np.cos(theta))

a = 100
theta = np.arange(0,2 * np.pi,0.01)
polar_unlabeled()
plt.title('Ellipses with Equal Maximum Radius and Common Focus')
for ecc in np.linspace(0,0.8,5):
    ecc = round(ecc,1)
    r = ellipse(theta, a, ecc)
    plt.plot(theta,r,label=("ε = "+str(ecc)))
plt.scatter(0,0,label='focus')
plt.legend()
plt.show()
```

If we align the ellipses to a common center, 
the eccentricity of 0.2 is barely visible. 
For comparison, the eccentricity of the earth's orbit is only 0.0167, 
making it even more remarkable that Kepler detected it.

```python
def ellipse(theta, a, eccentricity):
    return a * (1 - eccentricity**2) / (1 + eccentricity * np.cos(theta))
    
def ellipse_from_center(theta, a, eccentricity):
    dist = ellipse(theta, a, eccentricity)
    r_horiz = dist * np.cos(theta) + a * eccentricity
    r_vert = dist * np.sin(theta)
    r = np.sqrt(r_horiz**2 + r_vert**2)
    angle = np.angle(r_horiz + r_vert * 1j)
    return r,angle

a = 100
theta = np.arange(0,2 * np.pi,0.001)
polar_unlabeled()
plt.title('Ellipses with Equal Maximum Radius and Common Origin')
for eccentricity in np.linspace(0,0.8,5):
    ecc = round(eccentricity,1)
    r,angle = ellipse_from_center(theta, a, ecc)
    plt.plot(angle,r,label=("ε = "+str(ecc)))
plt.legend()
plt.ylim(bottom=0)    # this line needed for proper scaling
plt.show()
```

Let's check that the *x* and *y* coordinates measured from the center satisfy the equation
*x*<sup>2</sup> + *y*<sup>2</sup> /(1 - *eccentricity*<sup>2</sup>) = *a*<sup>2</sup>.

```python
a,ecc = 10,0.3
theta = np.arange(0,2 * np.pi,0.01)
print('Maximum relative divergence from ellipse equation in rectangular coordinates')
for ecc in np.linspace(0,0.8,5):
    r,angle = ellipse_from_center(theta, a, ecc)
    x = r * np.cos(angle)
    y = r * np.sin(angle)
    left_hand_side = x**2 + y**2 / (1 - ecc**2)
    log_ratio = np.log(left_hand_side) - 2 * np.log(a)
    max_diverge = max(np.abs(log_ratio))
    print('eccentricity','{:.1f}'.format(ecc),':','{:.0e}'.format(max_diverge))
```

Let's simulate gravitational orbits with any starting position (*r*,0) from a large mass and  any starting velocity (*vx,vy*). 
The code is adapted from Level 4 Part 3.

```python
def simulate_any_orbit(r,vx,vy,GM,num=2000,circuits=2):
    period_approx = 2 * np.pi * r**1.5 / np.sqrt(GM)
    agg_time = circuits * period_approx
    time = np.linspace(0,agg_time,num+1)
    x,y = np.zeros(num+1),np.zeros(num+1)
    dt = agg_time / num
    x[0] = curr_x = r
    curr_y = 0
    for i in range(1,num+1):
        curr_r = np.sqrt(curr_x**2 + curr_y**2)
        gravity = GM / curr_r**2
        avg_vx = vx - 0.5 * gravity * (curr_x / curr_r) * dt
        avg_vy = vy - 0.5 * gravity * (curr_y / curr_r) * dt
        dx = avg_vx * dt
        dy = avg_vy * dt
        avg_x = curr_x + 0.5 * dx
        avg_y = curr_y + 0.5 * dy
        avg_r = np.sqrt(avg_x**2 + avg_y**2)
        gravity = GM / avg_r**2
        vx -= gravity * (avg_x / avg_r) * dt
        vy -= gravity * (avg_y / avg_r) * dt
        x[i] = curr_x = curr_x + dx
        y[i] = curr_y = curr_y + dy
    return time,x,y

showscreen()
clearscreen()
r,vx,vy,GM = 140,8,8,20000
num,circuits = 2000,2
time,x,y = simulate_any_orbit(r,vx,vy,GM,num,circuits)
t = Turtle()
t.dot(10,'orange')
t.hideturtle()
skip = 3    # raise this to speed orbit
plot_xy(x[::3],y[::3],'purple')
```

Let's chop the orbit to a single circuit, 
fit an ellipse to it, and check that the maximum relative divergence is tiny. 
This is Kepler's First Law.

```python
def single_circuit(x,y):
    x_adj = (x - x[0]) * np.sign(x[1]-x[0])
    ix = 1
    while x_adj[ix] > 0: ix += 1
    while x_adj[ix] < 0: ix += 1
    x = x[:ix+1]
    y = y[:ix+1]
    return x,y

def ellipse_given_focus_at_origin(x,y):
    z = x + y * 1j
    r = np.abs(z)
    kmin = np.argmin(r)
    kmax = np.argmax(r)
    a = abs(z[kmin] - z[kmax]) / 2
    eccentricity = 1  - r[kmin] / a
    cos_major_axis = (z[kmin].real - z[kmax].real) / (2 * a)
    sin_major_axis = (z[kmin].imag - z[kmax].imag) / (2 * a)
    slope_major = sin_major_axis / cos_major_axis
    cosine = z.real / abs(z)
    sine = z.imag /abs(z)
    cosine_adj = cosine * cos_major_axis + sine * sin_major_axis
    r_ideal = a * (1 - eccentricity**2) / (1 + eccentricity * cosine_adj)
    rel_diverge = np.log(r_ideal/r)
    max_diverge = np.max(np.abs(rel_diverge))
    return a,eccentricity,slope_major,max_diverge

r,vx,vy,GM,num,circuits = 140,8,8,20000,2000,2
time,x,y = simulate_any_orbit(r,vx,vy,GM,num,circuits)
x,y = single_circuit(x,y)
a,eccentricity,slope_major,max_diverge = ellipse_given_focus_at_origin(x,y)
print('Best elliptical fit sets a =','{:.2f}'.format(a),', eccentricity =','{:.3f}'.format(eccentricity), \
      ', and slope of major axis =','{:.3f}'.format(slope_major))
print('Maximum relative divergence in distance r from gravitational focus is','{:.1e}'.format(max_diverge))
```

Let's compare the areas swept out around the orbit in equal time intervals. 
According to Kepler's Second Law, they should be equal. 
The first code below approximates the areas as triangles. 
The second code approximates the areas as slices with circular arcs.

```python
def compare_area_swept_out (x,y):    # triangular approx
    area = np.zeros(len(x)-1)
    for i in range(len(area)):
        # this formula is easily verified using geometry
        area[i] = 0.5 * (x[i+1] * y[i] - x[i] * y[i+1])
    area = np.abs(area)
    max_diverge = np.log(np.max(area) / np.min(area))
    return max_diverge

def compare_area_swept_out_alt(x,y):  # polar approx
    z = x + y * 1j
    r = np.abs(z)
    theta = np.angle(z)
    avg_r = 0.5 * (r[:-1] + r[1:])
    dtheta = np.abs(np.diff(theta))
    ix_max = np.argmax(dtheta)
    diff_max = dtheta[ix_max]
    if diff_max > 1.8 * np.pi:
        dtheta[ix_max] = np.abs(diff_max - 2 * np.pi)
    area = 0.5 * avg_r**2 * dtheta  # length avg_r and base of width dtheta
    max_diverge = np.log(np.max(area) / np.min(area))
    return max_diverge

r,vx,vy,GM,num,circuits = 140,8,8,20000,2000,2
time,x,y = simulate_any_orbit(r,vx,vy,GM,num,circuits)
x,y = single_circuit(x,y)
print('Maximum relative divergences in area swept out in time dt:') 
max_diverge = compare_area_swept_out(x,y)
print('{:.1e}'.format(max_diverge),'by triangular approximation')
max_diverge_alt = compare_area_swept_out_alt(x,y)
print('{:.1e}'.format(max_diverge_alt),'by polar approximation')
```

Let's check that the approximate orbital periods come close to satisfying Kepler's Third Law.

```python
def compare_orbital_period(r,vx,vy,GM,num=4000,circuits=4):
    time,x,y = simulate_any_orbit(r,vx,vy,GM,num,circuits)
    x,y = single_circuit(x,y)
    period = time[len(x)]
    a,eccentricity,slope_major,max_diverge = ellipse_given_focus_at_origin(x,y)
    period_ideal = 2 * np.pi * a**1.5 / np.sqrt(GM)
    rel_diverge = np.abs(np.log(period / period_ideal))
    return rel_diverge

sum_diverge = max_diverge = count = 0
for r in (200,250):
    for vx in (-4,-2,0,6,8):
        for vy in (-8,6,8):
            for GM in (20000,25000):
                rel_diverge = compare_orbital_period(r,vx,vy,GM)
                if rel_diverge > max_diverge:
                    max_diverge = rel_diverge
                count += 1
                sum_diverge += rel_diverge

print('Absolute relative divergences in estimated period from ideal prediction:')
print('Maximum','{:.1e}'.format(max_diverge),'   Mean','{:.1e}'.format(sum_diverge/count))
```
