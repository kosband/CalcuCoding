Use **print** to print multiple outputs from one cell. Note that we can keep changing width and length without getting error messages.
That's because "=" doesn't mean "equals" in Python (or most other programming languages). It means "replace with".

```python
# press the "Run" arrow above to see only print() results
perimeter = 14
width = 3
length = perimeter/2 - width
print(length)
width = 5                      # redefine width
length = perimeter/2 - width   # redefine length
print(length)
width = 1                      # redefine width
length = perimeter/2 - width   # redefine length
print(length)
```

Here's how to print several outputs on one line and add descriptions. 
Add single ' or double " quotation marks before and after any text you want to print.

```python
# calculate rectangle lengths given perimeter and various widths
perimeter = 14
width = 3
length = perimeter/2 - width
print('The width is',width,' and the length is',length) 
width = 5
length = perimeter/2 - width
print('The width is',width,'and the length is',length) 
width = 1
length = perimeter/2 - width
print('The width is',width,'and the length is',length)
```

Repeating instructions gets tedious. To simplify, we can use a **for** loop,
which contains an introductory statement and one or more indented lines below. Here is a simpler rewrite of the previous example.

```python
# calculate rectangle lengths given perimeter and various widths
perimeter = 14
for width in (3,5,1):
    length = perimeter/2 - width
    print('The width is',width,'and the length is',length)
```

Each run of the **for** loop is called an iteration. The start "for width" tells it to vary the width.
The continuation "in (3,5,1):" tells it to use 3,5,1 in that order. 
What else does each iteration evaluate? Whatever lines are indented just below the colon :.
If you forget the colon, you'll get an error message. 

```python
perimeter = 14
for width in (3,5,1)
    length = perimeter/2 - width
    print('The width is',width,'and the length is',length)
```

When you stop indenting, the **for** loop stops and the next line runs only once, using whatever values the **for** loop ended with.
Be careful to indent every line you want iterated and nothing else.

```python
perimeter = 14
for width in (3,5,1):
    length = perimeter/2 - width
print('The width is',width,'and the length is',length)
```

Since *x* never equals *x*+1, we never write *x* = *x* + 1 in ordinary algebra.
But in Python, *x = x* + 1 just means "change *x* from its current value to a value 1 higher".

```python
width = 1
print('width =',width)
width = width + 2
print('now width =',width)
```

Replacing values with related values is so common that Python provides shorthands for it: 
*x*+=*c* for *x* = *x* + *c*, *x*-=*c* for *x* = *x - c*,  _x_ *= _c_ for *x* = *x* * *c*, and *x*/=*c* for *x* = *x / c*.

```python
width = 5
print(width)
width += 2
print(width)
width -= 3
print(width)
width *=3
print(width)
width /= 4
print(width)
```
