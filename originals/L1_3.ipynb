Now let’s teach Python to measure the areas of polygons. 
For that we'll use the turtle function **pos**, which gives the current (*x,y*) position as a tuple. 
We can extract the *x* and *y* coordinates separately through indexing.

```python
%pip install ipycc
from ipycc.turtle import Turtle, showscreen, clearscreen
showscreen()
t = Turtle()
side = 100,200
turn = 130,0
t.reset()
p0 = t.pos()
t.forward(side[0])
p1 = t.pos()
t.left(turn[0])
t.forward(side[1])
p2 = t.pos()
t.home()
print('p0 =',p0,', p1 =', p1, 'p2 =', p2)
print('p1[0] = ',p1[0],', p2[0] = ',p2[0],', p2[1] = ',p2[1])
```

This code works for any polygon that starts at (0,0) or includes all sides.
Look up "shoelace formula" to learn more about it. 

Here *p* and *q* indicate positions at the start and end of an iteration. 
For starts from *p*=(0,0), the first *q* equals (side[0],0) and area stays 0. 
Then *p* shifts to *q*, *q* shifts to the position after left(turn[1]) 
and forward(side[1]), and area equals side[0] times *q*[1] divided by 2. 
With a triangle we’re done; otherwise, keep iterating.

```python
showscreen()
t.reset()
side = 100,200
turn = 90,0
# area computation given turtle import and tuples of side and turn
area = 0    # initial area
p = t.pos()   # p at start of side[0]
for k in range(len(side)):
    t.forward(side[k])
    q = t.pos()  # q is position at end of side[k]
    area += (p[0] * q[1] - p[1] * q[0]) / 2 
    p = q      # p shifts to position at start of side[k+1]
    t.left(turn[k])
t.home()
print('area =', area)
```

Here is an alternative expression of the previous **for** loop without numeric indices on *p* or *q*.

```python
t.home()
side = 100,200
turn = 90,0
# alternative expression of area computation
area = 0
px,py = t.pos()      # this assigns pos[0] to px and pos[1] to py
for k in range(len(side)):
    t.forward(side[k])
    qx,qy = t.pos()  # this assigns pos[0] to qx and pos[1] to qy
    area += (px * qy - py * qx) / 2
    px,py = qx,qy  # thus sets px=qx and py=qy 
    t.left(turn[k])
t.home()
print('area =', area)
```

Here is complete code to draw any polygon and its regular counterpart
with the same perimeter and compare the areas. 
It assumes that the initial position is (0,0) and 
that an unspecified last side returns to it. 
Note that Python lets us chain assignments; 
area=*px*=*py*=0 sets all three variables to 0.

```python
showscreen()
t.reset()
side = 75,100,100,75,50,150
turn = 40,60,70,30,40,0
# draw polygon, fill it, and compute perimeter and area
t.color('blue','cyan')
area = px = py = 0    # chained assignments with the same value
t.begin_fill()
for k in range(len(side)):
    t.forward(side[k])
    qx,qy = t.pos()
    area += (px * qy - py * qx) / 2
    px,py = qx,qy 
    t.left(turn[k])
perimeter = sum(side) + t.distance(0,0)
t.home()
t.end_fill()
# draw polygon w same perimeter and compute area
t.color('black')
num_sides = len(side) + 1
new_area = px = py = 0    # sets all three values to 0
for k in range(num_sides):
    t.forward(perimeter / num_sides)
    qx,qy = t.pos()
    new_area += (px * qy - py * qx)/2
    px,py = qx,qy
    t.left(360 / num_sides)
# print ratio of areas
ratio = round(new_area / area, 4)
print('The ratio of areas is', ratio)
```

Next, let us confirm that adding sides to a regular polygon increases the area for a given perimeter. 
This creates a **for** loop that iterates over the number of sides. 
Since each iteration draws and measures a polygon using a **for** loop, 
we need to nest one for loop inside another.

It is important to keep area=0 inside the outer loop. 
If outside, the area reported for a square will include the triangle’s area too,
that area will be included in the area reported for a pentagon, and so on. 
Putting statements outside a loop when they should be inside or 
vice-versa are common errors.

```python
showscreen()
t.reset()
# compare areas for polygons with same perimeter
perimeter = 500
print('# sides   area')
colors = ('maroon','blue','orange','green','turquoise','salmon','purple')
for n in range(3,10):   # range(3,10) is shorthand for 3,4,5,6,7,8,9
    area = px = py = 0
    t.color(colors[n-3])
    t.home()
    t.left(180 / n)  # half-turn makes polygon symmetric across y-axis
    for k in range(n):
        t.forward(perimeter / n)
        qx,qy = t.pos()
        area += (px * qy - py * qx) / 2
        px,py = qx,qy
        t.left(360 / n)
    print('   ', n,'  ',round(area))
```

To iterate over more sides, we can use **range**(start,stop,step). 
Its sequence runs (start, start+step, start+2 step,...) 
and continues until the next step would reach or surpass stop. 
Hence **range**(start,stop) is short for **range**(start,stop,1), 
while **range**(stop) is short for **range**(0,stop,1).

```python
showscreen()
t.reset()
# compare areas for many-sided polygons with same perimeter
perimeter = 500
print('# sides   area')
for n in range(10,100,10):   # shorthand for 10,20,30,40,50,60,70,80,90
    area = px = py = 0
    t.home()
    t.left(180 / n)  # half-turn makes polygon symmetric across y-axis
    for k in range(n):
        t.forward(perimeter / n)
        qx,qy = t.pos()
        area += (px * qy - py * qx) / 2
        px,py = qx,qy
        t.left(360 / n)
    print('  ', n,'  ',round(area))
```

Beyond 50 sides, all the polygons look circular. 
However, their area keeps rising nudge by nudge. 
The maximum possible area, which characterizes a perfect circle, 
is 500*500/4/𝜋 or approximately 19894.37. 
Our best attempt below falls short by one part in a million.

```python
showscreen()
t.reset()
# compare areas for extremely many-sided polygons with same perimeter
perimeter = 500
t.speed('fastest')  # this speeds up turtle but the code still runs slowly
# ok to press Stop above if really bored :-) 
print('# sides    area')
for n in range(100,600,200):   # shorthand for 100,300,500
    area = px = py = 0
    t.home()
    t.left(180 / n)  # half-turn makes polygon symmetric across y-axis
    for k in range(n):
        t.forward(perimeter / n)
        qx,qy = t.pos()
        area += (px * qy - py * qx) / 2
        px,py = qx,qy
        t.left(360 / n)
    print(' ', n,'  ',round(area,2))
```

The simplest way to store (sides,area) pairs for later reference is to form a **list**. 
A list looks like a tuple with elements enclosed in brackets [ ] rather than parentheses ( ). 
The core difference is that tuples can't be modified without complete replacement, whereas lists can. 
In Python terminology, tuples are *immutable* while lists are *mutable*.

```python
# elements can be changed in lists but not tuples
a_list = [2,3,6]
a_list[2] = 5  # modifies list to replace 6 with 5 
print(a_list)  # confirm the modification
a_tuple = (2.3,6)
a_tuple[2] = 5  # no replacement allowed for tuples so this will generate error
```

Python lists include a method **append** that adds elements at the end. 
We use it by typing the name of the list followed by '.append(name)'.

```python
# can add elements to lists but not to tuples
a_list = [2,3,5]
a_list.append(8)     # adds 8 to the list
print(a_list)        # confirm the additional element
# any of the following will generate errors
# append(a_list,8)   # append must come after the list name
# a_tuple(append(8)) # append must be separated with . not ()
# a_tuple.append(8)  # tuples can't be modified
```

IF *name* is a list or tuple, **append** adds it at a single index.
Use **extend** or the addition sign to add the elements one by one. 
To remove a list element, use either **remove** after the list name or **del** before.

```python
a_list = [2,3,5,8]
name1 = (9,10)
a_list.append(name1)  # appends the tuple as a single element
print(a_list)
name2 = (11,12)
a_list.extend(name2)  # adds 11 and 12 as two distinct elements
name3 = (13,14)
a_list += name3       # another way to add distint elements
print(a_list)
a_list.remove(12)     # removes the first element with value 12
print(a_list)
del a_list[4]         # removes the element at index 4 
print(a_list)
```

Let us redo our previous area calculations with the results stored as a list of lists. 

```python
showscreen()
t.reset()
perimeter = 500
t.speed('fastest')  # this will still be slow with many sides
# combine ranges into single list
list1 = list(range(3,10))   # converts range() into a list
list2 = list(range(10,100,10))
list3 = [100,300,500]  
num_sides = list1 + list2 + list3  # joins the lists together
# core calculations
results = [];  # creates empty list
for n in num_sides:   # shorthand for 100,400,700,1000,1300,1600,1900
    area = px = py = 0
    t.home()
    t.left(180 / n)  # half-turn makes polygon symmetric across y-axis
    for k in range(n):
        t.forward(perimeter / n)
        qx,qy = t.pos()
        area += (px * qy - py * qx) / 2
        px,py = qx,qy
        t.left(360 / n)
    results.append([n, round(area,2)]) 
print(results)
```

Next let us express the areas as percentages of the theoretical maximum 
and add those to the list of results. 
Note that to query the list, we need to use double indices, 
with the first index referring to a particular [*n,area*] list 
and the second index referring to the index within that list.

```python
max_area = perimeter**2 / 4 / 3.14159265459
print('# sides  ratio to max area')
for k in range(len(results)):
     ratio = results[k][1] / max_area  # area for this iteration is index 1 of results[k]
     results[k].append(ratio)          # add ratio to list          
     print('  ',results[k][0],'     ',round(results[k][2],5))
```

 More turtle art. Note the commands **break**, which terminates loops.

```python
showscreen()
t.reset()
side = 150,125
divisor = 12,10  # best to use only whole number factors of 180
t.speed('fastest')
t.color('red', 'yellow')
t.begin_fill()
while True:      # repeat forever until something inside this loop stops it
    t.forward(side[0])
    t.left(180-180/divisor[0])
    t.forward(side[1])
    t.left(180-180/divisor[1])
    p = t.pos()
    if abs(p[0])+abs(p[1]) < 1:  # checks for return near center
        break           # stops iterations after return near center
t.end_fill()
```
