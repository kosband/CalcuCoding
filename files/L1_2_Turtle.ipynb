# Introducing Python Turtle

The code below creates a turtle, a little figure that moves around a screen and draws as it goes. Below we call it "t". 

```python
%pip install ipycc   
from ipycc.turtle import Turtle,showscreen,clearscreen 
showscreen()         # opens a screen below this cell
t = Turtle()         # makes a turtle named t and places it at the center, pointing to the right. 
```

We tell Turtle "t" what to do by writing "t." just before the command. Every Turtle command needs parentheses, even when there is nothing between them.

```python
t.forward(100)       # t moves forward 100 steps (it can move 200 from the center in any direction)
t.left(90)           # t turns left 90 degrees, where 360 degrees turns full circle
t.forward(150)       # t moves forward 150 steps in the new direction 
```

Let's draw a rectangle with a purple border and a yellow interior. The turtle will gain a purple border and yellow interior too. 

```python
showscreen()       # display another screen view below this cell 
clearscreen()      # clear whatever was on the screen before
t = Turtle()
t.color("purple", "yellow")
t.begin_fill()     # start of the figure to fill
t.forward(100)     
t.left(90)         
t.forward(180)
t.left(90)
t.forward(100)     
t.left(90)         
t.forward(180)
t.end_fill()       # end of the figure to fill
```

Let's simplify the drawing instructions by using a **for** loop.

```python
showscreen()        
clearscreen()
t = Turtle()
t.color("purple", "yellow")
t.begin_fill() 
for side in [100, 180, 100, 180]:   # draws four sides
    t.forward(side)
    t.left(90)
t.end_fill()   
```

Any name for the turtle will do. Its shape and size can be changed, and the border color and fill color can be set separately.

```python
showscreen()        
clearscreen()
ninja = Turtle()
ninja.shape("turtle")     # other shapes are arrow, circle, square, triangle, and classic (the default)
ninja.shapesize(2, 2)     # makes turtle twice its normal size  
ninja.pencolor('navy')    # sets border color (default is black)
ninja.fillcolor('aqua')   # sets fill color (default is black)
ninja.begin_fill()  
for side in [180, 100, 180, 100]:  
    ninja.forward(side)
    ninja.left(90)
ninja.end_fill()    
```

Now create a different turtle. It moves a short distance left without drawing, then draws a narrower rectangle that overlaps the first.

```python
showscreen()
# clearscreen()            # don't clear screen as both rectangles belong in this picture
wizard = Turtle()
wizard.shape("triangle")
wizard.color("red", "pink")
wizard.shapesize(1.5, 1.5)
wizard.penup()            # raise pen so that it won't draw
wizard.goto(50, 150)
wizard.setheading(270)    # face down rather than to the right
wizard.pendown()          # lower pen so that it will draw
wizard.begin_fill()
for side in [200, 80, 200, 80]:
    wizard.forward(side)
    wizard.left(90)
wizard.end_fill()
```
