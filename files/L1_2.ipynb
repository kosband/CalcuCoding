Let's draw a triangle using turtle. To bring the turtle back to the center, use the command **home**.

```python
showscreen()
clearscreen()
t = Turtle()
t.color('blue','cyan')
t.begin_fill()
t.forward(150)  # takes 150 turtle steps forward
t.left(60)      # turn left 60 degrees
t.forward(100)  # takes 100 turtle steps forward in new direction
t.home()        # return to center and re-point to the right
t.end_fill()
```

To draw triangles of various shapes and sizes, we change the numeric inputs. 
For clarity and less risk of error, 
it is best not to do this in the middle of the code. 
Instead, let’s define variables for the two sides
and turning angle and move the settings to the top.

```python
showscreen()
side1 = 200
side2 = 100
turn = 130
t.color('red','pink')
# t.fillcolor('pink')
t.begin_fill()
t.forward(side1)
t.right(turn)
t.forward(side2)
t.home()
t.end_fill()
```

To clear previous turtle features without creating a new turtle, tell it to **reset**. 
We can also squeeze variable settings into one line, as Python interprets “x,y,z = 1,2,3” as “x=1, y=2, and z=3”.

```python
showscreen()
t.reset() 
side1,side2,turn = 200,100,130
# draw first triangle
t.forward(side1)
t.right(turn)
t.forward(side2)
t.home()
#t.reset()      # clear screen
# draw second triangle with colors
t.pencolor('brown')
t.fillcolor('yellow')
t.begin_fill()
t.forward(side1)
t.left(turn)
t.forward(side2)
t.home()
t.end_fill()
```

Next let’s calculate the triangle’s perimeter. 
Use **distance**(*x,y*) to indicate the turtle's distance to a point (*x,y*)
on the turtle screen, where the center is (0,0). 
If we compute “side3 = distance(0,0)” just before the turtle returns home, 
the perimeter equals side1+side2+side3.

```python
showscreen()
t.reset()
side1,side2,turn =  200,100,130
t.color('purple','plum')
t.begin_fill()
t.forward(side1)
t.left(turn)
t.forward(side2)
side3 = t.distance(0,0)
perimeter = side1 + side2 + side3
t.home()
t.end_fill()
print('The perimeter is',perimeter)
```

Let's superimpose an equilateral triangle with the same perimeter.

```python
t.color('black')         # when only one color is provided, fillcolor matches pencolor
equal_side = perimeter/3
t.forward(equal_side)
t.left(360/3)
t.forward(equal_side)
t.home()
```

Clearly, the equilateral triangle spans a larger area. How much larger? Ask the turtle. 
Given a start from (0,0), the height is the y-coordinate of the top corner, which the command **ycor** provides. 
Multiply by half the base to obtain the area. Do this for both triangles and compare.

```python
showscreen()
t.reset()
side1,side2,turn = 200,100,130
t.color('purple','plum')
t.begin_fill()
t.forward(side1)
t.left(turn)
t.forward(side2)
side3 = t.distance(0,0)
perimeter = side1 + side2 + side3
area = t.ycor() * side1 / 2
t.home()
t.end_fill()
# overlay an equilateral triangle with the same perimeter
t.color('black')
equal_side = perimeter/3
t.forward(equal_side)
t.left(360/3)
t.forward(equal_side)
new_area = t.ycor() * equal_side / 2
t.home()
ratio = new_area / area
print('The ratio of areas is',ratio)           # this will look messy
print('The ratio of areas is',round(ratio,4))  # round(x,n) rounds x to n significant digits
```

An underscore with small letters is the preferred Python convention for separating two words within a name. 
An alternative "camel" convention uses capital letters without underscores to start new words. 
Applying a convention consistently makes code more readable and less prone to error.

```python
perimeter = 400
equalSide = perimeter/3           # camel convention, with capital letters resembling humps
new_area = 0.866 * equalSide**2   # bad approach as it mixes two conventions
newArea = 0.866 * equalSide**2    # consistent camel convention
equal_side = perimeter/3          # preferred Python convention
new_area = 0.866 * equal_side**2  # preferred Python convention
print('New area is',new_area)
# next line is technically clearer, but long names can be distracting 
area_of_equilateral_triangle = 0.866 * equal_side**2 
print('New area is',area_of_equilateral_triangle)
```

Next let’s draw and measure quadrilaterals.

```python
showscreen()
t.reset()
side1,side2,side3 = 120,180,60
turn1,turn2 = 130,30
t.forward(side1)
t.left(turn1)
t.forward(side2)
t.left(turn2)
t.forward(side3)
t.home()
```

For neater expression, we can define side=100,150,50 to handle all three sides 
and turn=50,130 to handle both angles.
These multi-valued variables are known as tuples 
(with 'u' pronounced as either "uh" or "ooh").
To identify a particular value, use numbers enclosed in brackets [ ], 
where the first value is marked [0], the second value [1] and so on.

```python
showscreen()
t.reset()
side = 120,180,60
turn = 130,30
t.forward(side[0])
t.left(turn[0])
t.forward(side[1])
t.left(turn[1])
t.forward(side[2])
t.home()
```

The tuple form is easier to generalize to multiple sides 
and we can use Python loops to simplify. 
While we need to assign turn[2], its value doesn't matter 
since the turtle heads straight home after.

```python
showscreen()
t.reset()
side = 120,180,60
turn = 130,30,0
for k in (0,1,2):
    t.forward(side[k])
    t.left(turn[k])
t.home()
```

All Python indices start from 0. That's how we calculate birthdays: 
your first birthday was the second time you were alive on that date. 
Many countries number floors likewise,
with ground floor numbered 0 instead of 1. 
The good news is that Python numbers every sequence the same way, 
which makes it easy to get used to.

```python
x=(1,2,3,4,5)
print('x[0]=',x[0],', x[1]=',x[1],', x[4]=',x[4])
```

To compute the perimeter, we need to add various lengths, 
which we can do either iteratively or by using the Python method **sum**.

```python
# long way to sum sequence
x = (1,2,3,4,5)
sum_x = 0  
for anything in x:
    sum_x += anything
print('The sum of',x,'is',sum_x)
# short way to sum sequence
sum_x = sum(x)
print('The sum of',x,'is',sum_x)
```

Here is complete code for drawing a quadrilateral
and a square with the same perimeter.

```python
showscreen()
t,reset()
side = 120,180,60  # includes all sides except the last
turn = 130,30,0
t.pencolor('blue')
t.fillcolor('cyan')
t.begin_fill()
for k in (0,1,2):
    t.forward(side[k])
    t.left(turn[k])
perimeter = sum(side) + t.distance(0,0)  # last side length is distance(0,0)
t.home()
t.end_fill()
# draw square w same perimeter
t.color('black')
for k in (0,1,2,3):
    t.forward(perimeter/4)
    t.left(360/4)
```

To extend this to polygons with N sides, we can make use of two more commands. 
One is **range**(N) which iterates through the sequence 0,1,2,..,N-1. 
The other is **len**, which stands for length and counts the number of distinct elements. 
This code draws two octagons.

```python
showscreen()
t,reset()
side = 100,50,100,50,50,100,50  # length of last side is not specified
turn = 40,60,40,40,50,30,0
t.pencolor('blue')
t.fillcolor('cyan')
t.begin_fill()
num_side = len(side)  # number of elements in the sequence called side
for k in range(num_side):  # range(7) amounts to (0,1,2,3,4,5,6) except that sequence isn't stored
    t.forward(side[k])
    t.left(turn[k])
perimeter = sum(side) + t.distance(0,0)
t.home()
t.end_fill()
# draw regular polygon w same perimeter
t.color('black', 'black')
num_all_sides = num_side + 1  # count the last side too
for k in range(num_all_sides):
    t.forward(perimeter / num_all_sides)
    t.left(360 / num_all_sides)
```
