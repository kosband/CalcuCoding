Suppose we move in *n* equal steps from (0,0) to (*xmax,ymax*), with horizontal nudge *dx*=*xmax/n* and vertical nudge *dy*=*ymax/n*. 
The code below does this and draws a right triangle underneath. Since *dy/dx* is constant, the picture looks the same regardless of *n*, 
which the code renames as "steps".

```python
%pip install ipycc
from ipycc.turtle import Turtle, showscreen, clearscreen
clearscreen()
showscreen()
t = Turtle()

xmax,ymax,steps = 200,150,10   # max x position, max y position, and number of steps
dx,dy = xmax/steps, ymax/steps
t.fillcolor('light yellow')
t.begin_fill()
xnew = ynew = 0
for i in range(steps):
    xnew += dx
    ynew += dy
    t.goto(xnew, ynew)    # draw line from current postion pos() to (xnew,ynew)
t.goto(xnew,0)            # draw vertical line back to starting y value of 0
t.home()                  # draw horizontal line back to start at (0,0)
t.end_fill()
```

To fit stairs *dx* wide just outside the triangle, 
turtle at (*x,y*) moves *dy* vertically and then *dx* horizontally. 
The area under that step will equal (*y*+*dx*)*dx*.

```python
showscreen()
t.reset()
xmax,ymax,steps = 200,150,10   # max x position, max y position, and number of steps
dx,dy = xmax/steps, ymax/steps
t.color('orange')    # makes pencolor and fillcolor orange
t.begin_fill()
xnew = ynew = area = 0
for i in range(steps):
    ynew += dy       # ynew nudges up just before the area is computed
    area += ynew * dx   
    t.goto(xnew, ynew)
    xnew += dx
    t.goto(xnew, ynew)
t.goto(xnew,0)
t.home()
t.end_fill()
t.color('black')
t.goto(xmax,ymax)
t.goto(xmax,0)
t.home()
print('Area of upper staircase is',area)
```

To fit stairs *dx* wide just inside the triangle, 
turtle at (*x,y*) moves *dx* horizontally and then *dy* vertically. 
The area under that step will equal _y_\*_dx_.

```python
showscreen()
print('Area of upper staircase is',area)
# add lower staircase
t.color('brown')
t.begin_fill()
xnew = ynew = area = 0
for i in range(steps):
    xnew += dx
    area += ynew * dx   
    t.goto(xnew, ynew)
    ynew += dy         # ynew doesn't increase until after area computed
    t.goto(xnew, ynew)
t.goto(xnew,0)
t.home()   
t.end_fill()
print('Area of lower staircase is',area)
```

Let's store coordinates of the *x* and *y* positions along the way. 
They can be computed in several ways.

```python
xmax,ymax,steps = 200,150,10   # max x position, max y position, and number of steps
dx,dy = xmax/steps, ymax/steps
# long way to form x list
x = []
for i in range(steps+1):
    x.append(i * dx)
print('x is',x)

# short way to form x using list comprehension
xx = [i * dx for i in range(steps+1)]
print('Are x and xx the same list?',x==xx)

# two ways to form y 
y = [i * dy for i in range(steps+1)]  # will only work if we know dy before we know y
print('y is',y)
yy = [val * (ymax/xmax) for val in x]  # defines y as constant multiple of x
print('Are y and yy the same list?',y==yy)
```

Indices make clearer what is already evident in the graphs, 
namely that the higher area includes a rectangle of height *ymax* 
that the lower area does not.

```python
showscreen()
t.reset()
xmax,ymax,steps = 200,150,10
dx = xmax / steps
x = [dx * i  for i in range(steps+1)]
y = [(ymax/xmax) * val for val in x]  # defines y as constant multiple of x
# plot upper staircase and compute area
area_high = 0
t.fillcolor('orange')
t.begin_fill()
for i in range(steps):
    t.goto(x[i], y[i+1])
    t.goto(x[i+1], y[i+1])
    area_high += y[i+1] * dx
t.goto(xmax,0)
t.home()
t.end_fill()
# plot lower staircasea and compute area
area_low = 0
t.fillcolor('brown')
t.begin_fill()
for i in range(steps):
    t.goto(x[i+1], y[i])
    t.goto(x[i+1], y[i+1])
    area_low += y[i] * dx
t.goto(xmax,0)
t.home()
t.end_fill()
print(round(area_low), '< area of triangle <', round(area_high))
```

Here's a simpler way to compute areas using **sum( )**. 
The relative difference between the high and low areas works out to 2/steps. 
So we would need 200 steps to bound the areas to 1% accuracy.

```python
xmax,ymax,steps = 200,150,10
dx = xmax / steps
x = [dx * i  for i in range(steps+1)]
y = [(ymax/xmax) * val for val in x] 
area_low = sum(y[:-1]) * dx        # sum omits y[-1], which equals ymax 
area_high = sum(y[1:]) * dx        # sum omits y[0], which equals minimum y 
diff_area = area_high - area_low   # = ymax * dx = ymax * xmax/steps
area_mid =  0.5 * (area_low + area_high)
rel_diff = diff_area / area_mid    # = 2/steps 
print('Does the relative difference',rel_diff,'in areas equal 2/steps?',abs(rel_diff-2/steps)<1E-10)
```

We need a huge number of steps to make the relative difference insignificant. 
That's why it's better to use precise formulas if we can find them.

```python
xmax,ymax,steps = 200,150,10
while steps < 1E6:
    steps *= 10
    dx = xmax / steps
    x = [i * dx for i in range(steps+1)]
    y = [(ymax/xmax) * val for val in x]
    area_low = sum(y[:-1]) * dx  
    area_high = sum(y[1:]) * dx
    area_mid = 0.5 * (area_low + area_high)
    rel_diff = (area_high - area_low) / area_mid 
    print(steps,'steps with 2/steps',2/steps,'and relative difference',round(rel_diff,8))
```

Now let's apply the same approach to to bound the area of a quarter circle. 
We will need to compute square roots, 
which requires a command **sqrt()** from the package math. 
We will type "import math" and "math.sqrt()" to use it. 

```python
import math  # this contains the command sqrt() and a host of other mathematical functions
# sqrt(2)    # this will generate an error, because Python doesn't see sqrt in the main list
root2 = math.sqrt(2)  # Python looks in the loaded math package for sqrt, finds it, and applies it
print('The square root of 2 is',root2)
print('The stored value of π is',math.pi) # math.pi is float approximation to π
```

A circle around the origin with radius *r* is defined by points (*x,y*) 
satisfying  𝑥⋅𝑥 + 𝑦⋅𝑦 = 𝑟⋅𝑟.
We will focus on the quarter circle with negative *x* and positive *y*.
We will set each *x* an equal distance *dx* apart and approximate the 
quarter circle with polygons through the selected (*x,y*) and the origin.

```python
showscreen()
t.reset()
radius,steps = 195,40
dx = radius/steps
x = [-radius + i * dx for i in range(steps+1)]
y = [math.sqrt(radius * radius - val * val) for val in x]
# draw a polygon with those points and the origin as vertices
t.color('blue', 'cyan')
t.begin_fill()
for i in range(steps+1):
    t.goto(x[i],y[i])
t.home()
t.end_fill()
```

Next let's build staircases just inside and outside the quarter circle. 
Their areas bound the area 𝜋⋅𝑟⋅𝑟/4 of the quarter circle, and we can use this to bound  𝜋 .

```python
showscreen()
# nest staircases inside and outside quarter circle and use areas to bound π 
t.color('black')
t.goto(x[0],0)
for i in range(steps): # upper steps
    t.goto(x[i],y[i+1])
    t.goto(x[i+1],y[i+1])
t.home()
t.color('light pink')
t.begin_fill()
t.goto(x[0],0)
for i in range(steps): # lower steps
    t.goto(x[i+1],y[i])
    t.goto(x[i+1],y[i+1])
t.home()
t.end_fill()
# compute areas and convert to bounds on π
area_high = sum(y[1:]) * dx 
area_low = sum(y[:-1]) * dx 
pi_high = area_high * 4 / (radius * radius)
pi_low = area_low * 4 / (radius * radius)
print(pi_low,'< π <',pi_high)
```

Those bounds on 𝜋 are much looser than we computed earlier. 
While every quadrupling of steps improves them, convergence is slow.

```python
# repeatedly quadruple steps to tighten bounds on π 
radius,steps = 195,40
for iter in range(6):
    steps *= 4
    dx = radius/steps
    x = [-radius + i * dx  for i in range(steps+1)]
    y = [math.sqrt(radius * radius - val * val) for val in x]
    area_high = sum(y[1:]) * dx  
    area_low = sum(y[:-1]) * dx  
    pi_high = area_high * 4 / (radius * radius)
    pi_low = area_low * 4 / (radius * radius)
    rel_diff = (pi_high - pi_low) / math.pi
    # any terms with parentheses around them can be extended to multiple lines
    print(steps,'steps:',round(pi_low,7),'< π <',round(pi_high,7),   # continues on next line
                             'with relative difference',round(rel_diff,6))
```

Our earlier estimated bounds on π relied on turtle's turns and distance estimators. 
Let's compute lower bounds on π from the sums of distances 
between the vertices on the rim of the quarter circle. 
Again, they are much lower than the corresponding area-based estimates.

```python
# estimate π by comparing outer perimeter to radius for approximate quarter circle
radius,steps = 195,40
pi_est = []
print('steps   estimated π     too low by')
for iter in range(6):
    steps *= 4
    dx = radius/steps
    x = [-radius + i * dx  for i in range(steps+1)]
    y = [math.sqrt(radius * radius - val * val) for val in x]
    quarterperim = 0
    for i in range(steps):
        dy = y[i+1] - y[i]
        quarterperim += math.sqrt(dx*dx + dy*dy)
    pi_low = 2 * quarterperim / radius
    diff_pi = math.pi - pi_low
    print(steps,'   ',round(pi_low,8),'    ', round(diff_pi,8))
    pi_est.append([steps, round(pi_low,8), round(diff_pi,8)])
print('These answers stored as list of lists')
print(pi_est)
```

Before closing this lesson, let's discuss a potential problem in working with lists. 
Suppose we want to make a backup copy of *list1*. 
If we write *backup*=*list1* and then change *list1*, *backup* changes too. 
That's because Python points the name *backup* to whatever *list1* points to and treats the two names as synonyms. 
This can be checked by using the Python query **is**, which is stronger than just checking for equal values.

```python
list1 = [[1, 2, 3], [4, 5, 6]]
backup = list1
list1[0][0]=10
print('When list1[0][0] is changed to',list1[0][0],', backup[0][0] changes to',backup[0][0])
backup[1][2]=20
print('When list1[1][2] is changed to',list1[1][2],', backup[1][2] changes to',backup[1][2])
print('Do all the elements of list1 and backup match in value?',list1==backup)
print('Do list1 and backup point to the same object?',list1 is backup)
```

The safest way to make a full copy is 
to import the package **copy** and use the command **copy.deepcopy()**.

```python
import copy
list1 = [[1, 2, 3], [4, 5, 6]]
backup = copy.deepcopy(list1)
print('Do all the elements of list1 and backup match in value?',list1==backup)
print('Do list1 and backup point to the same object?',list1 is backup)
list1[0][0] = backup[1][2] = 'X'
print('Index  list1  backup')
print('[0][0]   ',list1[0][0],'    ',backup[0][0])
print('[1][2]   ',list1[1][2],'    ',backup[1][2])
```

Tuples can be assigned multiple names too,

```python
tuple1 = (1,2,3)   
tuple2 = tuple1 
print('  tuple1   tuple2')
print(tuple1, tuple2)
tuple1 = (4,5,6)
print(tuple1, tuple2)
tuple2 = (7,8,9)
print(tuple1, tuple2)
```
