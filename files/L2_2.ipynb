Let's assume that velocity *v* is linear in time, and plot (*time,v*) using turtle. 
The net displacement equals the signed area between *v* and the *time*-axis.

```python
%pip install ipycc
from ipycc.turtle import Turtle, showscreen, clearscreen
showscreen()
clearscreen()
t = Turtle()

intercept,slope = 50,0.75
tmin,tmax,num_intervals = -200,200,20         # steps renamed as num_intervals 
dt = (tmax - tmin) / num_intervals
time = [tmin + i * dt for i in range(num_intervals+1)]
v = [intercept + slope * val for val in time]
t.color('blue','cyan')
t.begin_fill()
t.goto(time[0],0)
for i in range(num_intervals+1): 
    t.goto(time[i],v[i])
t.goto(time[-1],0)
t.home()
t.end_fill()
```

Every vertical slice creates a trapezoid, with area equal to base times average height. 
The only tricky part in calculus is that the height can be negative. 
But that serves us well in computing net displacement.

```python
showscreen()
# fill vertical slices and compute area
indices = 3,round(0.55*num_intervals),round(0.8 * num_intervals)
t.color('light pink')
for i in indices:
    t.penup() 
    t.goto(time[i],0)
    t.pendown() 
    t.begin_fill()
    t.goto(time[i],v[i])
    t.goto(time[i+1],v[i+1])
    t.goto(time[i+1],0)
    t.end_fill()
    area = (time[i+1]-time[i]) * 0.5 * (v[i]+v[i+1])
    print('area at step',i,'is',area )
```

Do you find the area computation confusing? For more intuition, stack two identical trapezoids with heights *v* 
on one side and *v*+*dv* on the other to create a rectangle with the same base times twice the average height.

```python
showscreen()
t.reset()
t.color('red','lightpink')
t.begin_fill()
t.goto(50,50)
t.write('  v + dv')   # turtle can write
t.goto(50,0)
t.write('  v')
t.goto(50,-150)
t.write('  0')
t.goto(0,-150)
t.home()
t.end_fill()
t.color('red','hotpink')
t.begin_fill()
t.goto(0,175)
t.goto(50,175)
t.write('  2v + dv')
t.goto(50,50)
t.home()
t.end_fill()
for i in range(10):  # make horizontal dashed line
    t.forward(2)
    t.penup() 
    t.forward(3)
    t.pendown()
```

Let's fill these slices one by one and plot the corresponding cumulative areas, 
which equal the net displacements at the various times. 
We'll need to rescale the area plot to keep it from running off the turtle canvas.

```python
showscreen()
t.reset()
# next line sets scale that usually keeps area plot on canvas
scale = 1000 /(time[-1] - time[0]) / abs(v[-1] - v[0])
curr_pos = (time[0],0)
cum_area = 0
for i in range(num_intervals):
    # uses code from previous trinket to fill slice
    t.penup()
    t.goto(time[i],0)
    t.color('red','lightpink')
    t.pendown()
    t.begin_fill()
    t.goto(time[i],v[i])
    t.goto(time[i+1],v[i+1])
    t.goto(time[i+1],0)
    t.end_fill()
    # update cumulative area and plot it
    cum_area += dt * 0.5 * (v[i]+v[i+1])
    t.penup()
    t.goto(curr_pos)
    t.color('black')
    t.pendown()
    curr_pos = (time[i+1], cum_area * scale)
    t.goto(curr_pos)
```

Let's compute and plot areas for various straight lines, 
without the tedium of filling slices one by one. 
The relevant code combines three main tasks.

```python
showscreen()
t.reset()
# Task 1: assign time and v values 
intercept,slope = 50,0.75
tmin,tmax,num_intervals = -200,200,20
dt = (tmax - tmin) / num_intervals
times = [tmin + i * dt for i in range(num_intervals+1)]
v = [intercept + slope * val for val in times]

# Task 2: chart (times,v) and area between v and times-axis 
t.color('blue','cyan')
t.begin_fill()
t.goto(time[0],0)
for i in range(num_intervals+1): 
    t.goto(time[i], v[i])
t.goto(time[-1],0)
t.home()
t.end_fill()

# Task 3: compute cumulative areas and chart rescaled values
t.penup()
t.goto(time[0],0)
t.color('black')
t.pendown()
scale = 1000 /(time[-1] - time[0]) / abs(v[-1] - v[0])
cum_area = 0
for i in range(num_intervals):
    cum_area += dt * 0.5 * (v[i]+v[i+1])
    t.goto(time[i+1], cum_area * scale)
```

To make code simpler, easier to follow, and less prone to error, it's often helpful to rewrite key tasks as Python **functions**. 
A function is a kind of machine that converts inputs into outputs. Here is a simple example.

```python
# first line must start with "def", specify name and inputs, and end with colon
def drawSquare(side): 
    t = Turtle()        # all lines after the first must be indented
    for k in range(3):
        t.forward(side)
        t.left(90)
    t.home()
    
# demonstrate its use ("demo"), which also checks that the code works
showscreen()
clearscreen()
drawSquare(50)
drawSquare(100)
drawSquare(150)
```

Let's convert Task 2 into a function and demo it. Note that input names inside a function don't need to match what they're called outside.

```python
def chart_area(x,y):          # convert Task 2
    t = Turtle()
    t.color('blue','cyan')     
    t.begin_fill()
    t.goto(x[0],0)
    for i in range(len(x)):   # len(x) helps identify the relevant range
        t.goto(x[i], y[i])
    t.goto(x[-1],0)
    t.end_fill()
    t.home()
    
# demo
showscreen()
clearscreen()
chart_area(time,v)   # this substitutes time for x and v for y
```

Next we'll convert Task 3. Beware variables that aren't explicit inputs or defined inside, since they might take on a value you didn't intend without your realizing it.

```python
def chart_cum_area(x,y):    # convert Task 3
    t = Turtle()
    t.penup()
    t.goto(x[0],0)
    t.color('black')
    t.pendown()
    scale = abs(1000 / (x[-1] - x[0]) / (y[-1] - y[0]))
    cum_area = 0
    for i in range(len(x)-1):
        # cum_area += dx * 0.5 * (y[i] + y[i+1])             # avoid, as dx not identified and might not be constant
        cum_area += (x[i+1] - x[i]) * 0.5 * (y[i] + y[i+1])  # guaranteed to work
        t.goto(x[i+1], cum_area * scale)

# demo
showscreen()
chart_cum_area(time,v)
```

To convert Task 1 into a function, the latter needs to output the lists of (*time,v*). Like inputs, outputs can have different names inside the function and outside.

```python
# convert Task 1 into a function
def create_time_velocity(intercept,slope,num_intervals):
    tmin,tmax = -200,200
    dt = (tmax - tmin) / num_intervals
    t = [tmin + i * dt for i in range(num_intervals+1)]   # here t is a list, even though it is a turtle outside
    v = [intercept + slope * val for val in t]  # defines v=a+b*t
    return t,v  # this generates t,v as outputs

# demo
intercept,slope,n = 50,0.75,40
time,v = create_time_velocity(intercept,slope,n)  
print(time[0],time[-1],v[0],v[-1], len(time), len(v))
```

Function definitions can assign default values to inputs, 
which are used whenever the inputs aren't specified. 
Input order can be changed provided the inputs are explicitly named.

```python
# convert Task 1 into a function with default value 40 for num_intervals
def create_time_velocity(intercept,slope,num_intervals=40):
    tmin,tmax = -200,200
    dt = (tmax - tmin) / num_intervals
    t = [tmin + i * dt for i in range(num_intervals+1)]
    v = [intercept + slope * val for val in t]  # defines v=a+b*t
    return t,v  # this generates time,velocity as outputs

# demo
time,v = create_time_velocity(num_intervals=60,slope=0.75,intercept=100)
print(time[0],time[-1],v[0],v[-1],'num_intervals is',len(time)-1)
time,v = create_time_velocity(slope=0.75,intercept=100)
print(time[0],time[-1],v[0],v[-1],'num_intervals is',len(time)-1)
```

Functions can contain other functions. That makes it easer to change parts without rewriting the whole.

```python
def chart_time_velocity_area(intercept,slope,num_intervals=40): 
    time,v = create_time_velocity(intercept,slope,num_intervals)
    chart_area(time,v)
    chart_cum_area(time,v)

# demo
showscreen()
clearscreen()
chart_time_velocity_area(-10,1,60)
```

We can use cumulative area to scale the chart more accurately. For more clarity, separate area computation from charting. 
Such changes are known as **refactoring**. Refactoring is as vital to good coding as editing is to good writing.

```python
def calc_cum_area(x,y):
    cum_area_i = 0
    cum_area = [0]    
    for i in range(1,len(x)):
        cum_area_i += (x[i] - x[i-1]) * 0.5 * (y[i-1] + y[i])
        cum_area.append(cum_area_i) 
    return cum_area  

def chart_cum_area(x,cum_area):   # better computation of scale
    cum_area_min = min(cum_area)   # Python function for minimum of a sequence
    cum_area_max = max(cum_area)   # Python function for maximum of a sequence
    scale = 200 / (cum_area_max - cum_area_min) 
    t = Turtle()  
    t.penup()        # here t is a turtle
    t.goto(x[0],0)
    t.pendown()
    for i in range(1,len(x)):
        t.goto(x[i], cum_area[i] * scale)

def chart_time_velocity_area(intercept,slope,num_intervals=40): 
    t,v = create_time_velocity(intercept,slope,num_intervals) 
    chart_area(t,v)  # here t is a list
    cum_area = calc_cum_area(t,v)
    chart_cum_area(t,cum_area)
    return t,v,cum_area 

# demo
showscreen()
clearscreen()
time,v,y = chart_time_velocity_area(-10,1,60)
```
