Let's have Python compute the area of a rectangle given its perimeter and width.

```python
perimeter = 14
width = 3
length = perimeter/2 - width
area = width * length
print('When width is',width,', area is',area)
```

To compare areas for different widths, let's use a **for** loop.

```python
perimeter = 14
for width in (0,0.5,1,1.5,2,2.5,3):
    length = perimeter/2 - width
    print('When width is',width,', area is',width * length)
```

Another kind of loop, called a **while** loop, 
proceeds until a condition listed just after the **while** is false. 
Like **for** loops, **while** loops end their first lines with a colon and indent the rest. 
Here is a **while** way to generate the same results as before.

```python
# calculate rectangle areas for perimeter and various widths
perimeter = 14
nudge = 0.5
width = 0
length = perimeter/2 - width
while width < length:
    print('When width is',width,', area is',width * length)
    width = width + nudge    # "=" means "replace with", not "equals"
    length = length - nudge
```

Now use the Python shorthands *x*+=*c* for *x*=*x*+*c* and *x*-=*c* 
for *x*=*x-c*. And if we want to keep nudging width until it equals length, 
we can change the **while** condition to "width<=length".

```python
# calculate rectangle areas for perimeter and various widths
perimeter = 14
nudge = 0.5
width = 0
length = perimeter/2 - width
while width <= length:
    print('When width is',width,', area is',width * length)
    width += nudge    # same as width = width + nudge
    length -= nudge   # same as length = length - nudge
```

A relation like "width<=length" is called a logical variable, with a value True or False. 
To test for equality, Python uses == for "equals" and != for "not equals". 
Logical variables can have names, and we can reverse their values by using 
the logical operator **not** or **not()**.

```python
print('1<2 is',1<2,', 3>2 is',3>2,', 4>=4 is',4>=4,', 5==5 is',5==5)
print('1>2 is',1>2,', 3<2 is',3<2,', 5<=4 is',5<=4,', 5!=5 is',5!=5)
print('True==True is',True==True,', False==False is',False==False,', True==Faslse is',True==False)
feature = (4<=3)      # parentheses not essential but help to keep things straight
print(feature, not(feature), not feature, not not feature, not(not(feature)))
```

Two other logical operators are **or** and **and**, 
which combine two logical variables into one.
'*A* or *B*' is True unless both *A* and *B* are False.
'*A* and *B*' is False unless both *A* and *B* are True.
'*A* > *B* or *A*==*B*' is equivalent to '*A*>=*B*'.
 A chain of logical tests should be read as repeated **and** statements.

```python
print("(1<2 or 2>3) is",1<2 or 2>3,", (1<2 and 2>3) is",1<2 and 2>3)
print("1<2 and 2<3) is",1<2 and 2<3,', 1<2>3 is',1<2>3)
print('1<2<3<4<5 is',1<2<3<4<5,', 1==2-1==3-2 is',1==2-1==3-2)
```

Tests for equality work great when comparing integers or logical values. Python also knows that the integer 9 equals the float 9.0 and so on. However, tests for float equality can be unreliable. Why? Rounding errors. Moreover, since computers do arithmetic in base 2, the errors can look inconsistent.

```python
nudge = 0.2
x = 2 * nudge
y = nudge + nudge
print('x =',x,', y =',y)
x = 2 + 2 * nudge; 
y = 2 + nudge + nudge;
print('x =',x,', y =',y)
x = 8 + 2 * nudge; 
y = 8 + nudge + nudge;
print('x =',x,', y =',y)
```

When we subtract two numbers that differ by small nudges, the nudges sometimes get modified or erased. 
Here 2e-14 or 2E-14 is scientific notation for 2⋅10**(−14), 
or a 2 placed 14 digits to the right of the decimal point.

```python
nudge = 2e-14
nudge1 = (5 + nudge) - 5
nudge2 = (500 + nudge) - 500
print(nudge, nudge1, nudge2)
```

How tiny can a nonzero float get? Let's halve a positive nudge as long as we can.

```python
nudge = 1
while nudge > 0:
    base = nudge
    nudge /= 2   # nudge = nudge / 2
print(base)
```

In practice, rounding errors tend to kick in well before the lower bound on nudge size.

```python
base = 1
divisor = 5
nudge = base / divisor
while (base == divisor * nudge):
    print(base)
    base = nudge
    nudge /= divisor
print(base, divisor * nudge)
```

Hence, the safest way to test two floats for equality is to allow some tiny difference between them. 
This code uses the Python function **abs** to take the absolute value.

```python
base = 1
divisor = 5
tiny = 1E-10
nudge = base / divisor
while (base == divisor * nudge):
    base = nudge
    nudge /= divisor
print('Is relative difference less than',tiny,'?')
while (base > tiny):
    diff = base - divisor * nudge
    print(base, diff, abs(diff)<base*tiny)
    base = nudge
    nudge /= divisor
```

Now let's return to the problem of maximizing 
rectangle area *A* by nudging the width.
Suppose we start with width=0 and nudge it by +1 until *A* declines.
Then let's nudge width by -0.4 until *A* declines,
then nudge it by 0.16 until *A* declines, and so on, 
with each switch multiplying the nudge by -0.4. 
Here's how we might code the process through the first switch in nudge.

```python
perimeter = 14
nudge = 1
reduce_nudge = -0.4  # must lie between 0 and -1
width = area = 0     # initial width and area
# first block of calculations
prev_area = -1  # ensure that initial nudge accepted
while area > prev_area:
    width += nudge # add nudge to width
    prev_area = area  # replace previous area with current area
    area = width * (perimeter/2 - width) # calculate new area
    print('Width is',width,', area is',area,', nudge is',nudge)
nudge *= reduce_nudge  # reduce nudge and switch sign
```

We can now add blocks to continue the process.

```python
perimeter = 14
nudge = 1
reduce_nudge = -0.4  # must lie between 0 and -1
width = area = 0     # initial width and area
# first block
prev_area = -1
while area > prev_area:
    width += nudge
    prev_area = area
    area = width * (perimeter/2 - width)
    print('Width is',width,', area is',area,', nudge is',nudge)
nudge *= reduce_nudge 
# second block
prev_area = area - 1
while area > prev_area:
    width += nudge
    prev_area = area
    area = width * (perimeter/2 - width)
    print('Width is',width,', area is',area,', nudge is',nudge)
nudge *= reduce_nudge
# third block
prev_area = area - 1
while area > prev_area:
    width += nudge
    prev_area = area
    area = width * (perimeter/2 - width)
    print('Width is',width,', area is',area,', nudge is',nudge)
nudge *= reduce_nudge
```

While we can keep adding blocks, the code is getting unwieldy and we don't know how many blocks we need. 
Here's another approach using the Python statement "**if** *Condition*:", 
which runs the indented code block below it when *Condition* is true and skips the block otherwise.

```python
perimeter = 14
nudge = 1
reduce_nudge = -0.4
width = prev_area = 0
num_steps = 0                   # initialize the number of steps
while num_steps < 15:           # stop for num_steps>=15
    num_steps += 1              # add 1 to num_steps
    area = width * (perimeter/2 - width)
    # next two lines reduce nudge and switch sign if and only area<=prev_area
    if area <= prev_area:       # don't forget the colon
        nudge *= reduce_nudge
    width += nudge 
    prev_area = area
    print('Width is',width,', area is',area,', nudge is',nudge)
```

Instead of restricting the number of steps, we can iterate until the changes are tiny.

```python
# search for maximum rectangular area through directed nudges
# initializations
perimeter = 14
nudge = 0.1
reduce_nudge = -0.4
tiny = 1E-16
width = 0
prev_area = -100
num_steps = 0
# core calculations
while abs(nudge) > tiny:
    num_steps += 1
    area = width * (perimeter/2 - width)
    if area <= prev_area:
        nudge *= reduce_nudge
    width += nudge
    prev_area = area
print('Width is',width,', area is',area,', number of steps is',num_steps)
```
