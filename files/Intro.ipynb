To raise *A* to the power (exponent) *B*, start with *A* at step 1, 
multiply by *A* at step 2, 
and keep multiplying by *A* for *B* steps. 
Later we will learn to raise to fractional and negative powers.

```python
5**2       # raise A to the power B by writing A**B
5**2 + 1   # equivalent to (5 * 5) + 1
5**(2+1)   # = 5**3 = 5 * 5 * 5
```

Note that we can keep changing width and length without getting error messages.
That's because "=" doesn't mean "equals" in Python (or most other programming languages).
It means "replace with".

```python
perimeter = 14
width = 3
length = perimeter/2 - width
print('The length is',length)
width = 5                      # redefine width
length = perimeter/2 - width   # redefine length
print('Now length is',length)
width = 1                      # redefine width
length = perimeter/2 - width   # redefine length
print('Now length is',length)
```

Here's how to print several outputs on one line and add descriptions. 
Add single ' or double " quotation marks before and 
after any text you want to print.

```python
# calculate rectangle lengths given perimeter and various widths
perimeter = 14
width = 3
length = perimeter/2 - width
print('When width is',width,', length is',length) 
width = 5
length = perimeter/2 - width
print('When width is',width,', length is',length) 
width = 1
length = perimeter/2 - width
print('When width is',width,', length is',length) 
```

Repeating instructions gets tedious. To simplify, we can use a **for** loop,
which contains an introductory statement and one or more indented lines below.
Here is a simpler rewrite of the previous example.

```python
# calculate rectangle lengths given perimeter and various widths
perimeter = 14
for width in (3,5,1):
    length = perimeter/2 - width
    print('When width is',width,', length is',length)
```

Each run of the **for** loop is called an iteration. The start "for width" tells it to vary the width.
The continuation "in (3,5,1):" tells it to use 3,5,1 in that order. 
What else does each iteration evaluate? Whatever lines are indented just below the colon :.
If you forget a needed ":", you'll get an error message.

```python
perimeter = 14
for width in (3,5,1)
    length = perimeter/2 - width
    print('When width is',width,', length is',length)
```

When you stop indenting, the **for** loop stops and the next line runs only once,
using whatever values the **for** loop ended with.
Be careful to indent every line you want iterated and nothing else.

```python
perimeter = 14
for width in (3,5,1):
    length = perimeter/2 - width
print('When width is',width,', length is',length)
```

Since *x* never equals *x*+1, we never write *x*=*x*+1 in ordinary algebra.
But in Python, *x=x*+1 just means "change *x* from its current value to a value 1 higher".

```python
width = 1
print('width =',width)
width = width + 2
print('Now width =',width)
```

Replacing values with related values is so common
that Python provides shorthands for it: 
*x*+=*c* for *x*=*x*+*c*, *x*-=*c* for *x*=*x-c*, 
_x_\*=_c_ for _x_=_x_\*_c_, and *x*/=*c*  for *x*=*x/c*.

```python
width = 5
print(width)
width += 2
print('Add 2:', width)
width -= 3
print('Subtract 3:', width)
width *=3
print('Multiply by 3:', width)
width /= 4
print('Divide by 4:', width)
```
